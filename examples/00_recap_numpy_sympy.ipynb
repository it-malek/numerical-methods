# 00 — NumPy / SymPy / Matplotlib recap

A short tour of the three libraries that underpin every lab in this collection.
This notebook is purely a setup check — running it confirms your environment
can produce numerical arrays, symbolic expressions, and plots.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({"figure.figsize": (7, 4.5), "figure.dpi": 110, "axes.grid": True})

import sympy as sp

x = np.linspace(-2, 2, 200)
fig, ax = plt.subplots()
ax.plot(x, np.sin(x), label=r"$\sin(x)$")
ax.plot(x, np.cos(x), label=r"$\cos(x)$")
ax.plot(x, x * np.exp(-(x**2)), label=r"$x e^{-x^2}$")
ax.legend(); ax.set_title("Three smooth functions")
fig.tight_layout(); plt.show()
```

## SymPy: symbolic differentiation and integration

```python
z = sp.Symbol('z')
expr = sp.exp(-z**2)
print('d/dz exp(-z^2) =', sp.diff(expr, z))
print('antiderivative =', sp.integrate(expr, z))
```

## Boolean masking with NumPy

```python
arr = np.array([-3, -1, 0, 1, 4, 9])
print('positive entries:', arr[arr > 0])
```
